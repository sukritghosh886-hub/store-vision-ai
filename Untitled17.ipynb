```python

# Store Vision AI — Synthetic Shelf Test

import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

BASE = "/content/store_vision_test"
os.makedirs(BASE, exist_ok=True)

W, H = 1280, 720
img = np.ones((H, W, 3), dtype=np.uint8) * 245

# Shelf background
cv2.rectangle(img, (80, 80), (1200, 640), (210, 210, 210), -1)

# Shelf levels
for y in [250, 430, 610]:
    cv2.rectangle(img, (80, y), (1200, y + 12), (80, 80, 80), -1)

# Generate exactly 27 synthetic products
product_colors = [
    (80, 120, 220),
    (100, 190, 100),
    (220, 120, 100),
    (180, 100, 200),
    (80, 180, 200),
    (200, 180, 80),
]

boxes = []
product_id = 0

for y1, y2 in [
    (105, 235),
    (275, 415),
    (455, 595)
]:
    xs = np.linspace(110, 1110, 9).astype(int)

    for x1 in xs:
        x1 = int(x1)
        x2 = x1 + 85

        color = product_colors[product_id % len(product_colors)]

        cv2.rectangle(
            img,
            (x1, y1),
            (x2, y2),
            color,
            -1
        )

        cv2.rectangle(
            img,
            (x1, y1),
            (x2, y2),
            (40, 40, 40),
            2
        )

        cv2.putText(
            img,
            str(product_id + 1),
            (x1 + 20, y1 + 65),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (255, 255, 255),
            2
        )

        boxes.append((x1, y1, x2, y2))
        product_id += 1

synthetic_path = f"{BASE}/synthetic_shelf.jpg"
cv2.imwrite(synthetic_path, img)

print("✅ Synthetic shelf created")
print("Ground-truth products:", len(boxes))
print("Image:", synthetic_path)

plt.figure(figsize=(16, 8))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python

# Store Vision AI — Advanced Synthetic Dataset
# Corrected version — no Kaggle required

import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

DATASET = "/content/store_vision_test/advanced"
os.makedirs(DATASET, exist_ok=True)

np.random.seed(42)

def create_synthetic_store_image(
    image_id,
    num_products=20,
    image_size=(720, 1280)
):
    H, W = image_size

    # Background
    img = np.ones((H, W, 3), dtype=np.uint8) * 245

    # Shelf
    cv2.rectangle(
        img,
        (50, 50),
        (W - 50, H - 50),
        (220, 220, 220),
        -1
    )

    # Shelf levels
    for y in [230, 450]:
        cv2.rectangle(
            img,
            (50, y),
            (W - 50, y + 10),
            (70, 70, 70),
            -1
        )

    # FIX: use a normal Python list and select by integer index
    shelf_ranges = [
        (70, 210),
        (270, 430),
        (490, 610)
    ]

    annotations = []

    for i in range(num_products):

        # Random product dimensions
        pw = int(np.random.randint(45, 100))
        ph = int(np.random.randint(80, 150))

        # Random shelf
        shelf_index = int(
            np.random.randint(0, len(shelf_ranges))
        )

        shelf_y1, shelf_y2 = shelf_ranges[shelf_index]

        # Make sure product fits vertically
        max_y = shelf_y2 - ph

        if max_y <= shelf_y1:
            max_y = shelf_y1 + 1

        x1 = int(
            np.random.randint(
                70,
                W - pw - 70
            )
        )

        y1 = int(
            np.random.randint(
                shelf_y1,
                max_y
            )
        )

        x2 = x1 + pw
        y2 = y1 + ph

        # Random product appearance
        color = tuple(
            int(x)
            for x in np.random.randint(
                50,
                230,
                size=3
            )
        )

        # Product body
        cv2.rectangle(
            img,
            (x1, y1),
            (x2, y2),
            color,
            -1
        )

        # Product border
        cv2.rectangle(
            img,
            (x1, y1),
            (x2, y2),
            (40, 40, 40),
            2
        )

        # Product label
        cv2.putText(
            img,
            f"P{i+1}",
            (
                x1 + 5,
                y1 + min(35, ph - 5)
            ),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (255, 255, 255),
            2
        )

        annotations.append({
            "id": i + 1,
            "x1": x1,
            "y1": y1,
            "x2": x2,
            "y2": y2,
            "width": pw,
            "height": ph
        })

    path = f"{DATASET}/store_{image_id}.jpg"

    success = cv2.imwrite(path, img)

    if not success:
        raise RuntimeError(
            f"Could not save image: {path}"
        )

    return path, annotations


# Generate 5 lightweight synthetic test images
dataset = []

for image_id in range(1, 6):

    number_of_products = int(
        np.random.randint(15, 31)
    )

    path, annotations = create_synthetic_store_image(
        image_id=image_id,
        num_products=number_of_products
    )

    dataset.append({
        "image": path,
        "ground_truth": len(annotations),
        "annotations": annotations
    })


print("=" * 50)
print("ADVANCED SYNTHETIC DATASET")
print("=" * 50)

print("Images created:", len(dataset))

for item in dataset:
    print(
        f"{os.path.basename(item['image'])} "
        f"→ {item['ground_truth']} products"
    )

print("\n✅ Synthetic dataset generation PASSED")
```

```python

# Store Vision AI — Actual YOLO Detection Evaluation

from ultralytics import YOLO
import numpy as np
import pandas as pd

# Load the lightweight pretrained model
model = YOLO("yolo11n.pt")

def calculate_iou(box_a, box_b):
    """
    Calculate Intersection over Union (IoU)
    for [x1, y1, x2, y2] boxes.
    """

    ax1, ay1, ax2, ay2 = box_a
    bx1, by1, bx2, by2 = box_b

    inter_x1 = max(ax1, bx1)
    inter_y1 = max(ay1, by1)
    inter_x2 = min(ax2, bx2)
    inter_y2 = min(ay2, by2)

    inter_w = max(0, inter_x2 - inter_x1)
    inter_h = max(0, inter_y2 - inter_y1)

    intersection = inter_w * inter_h

    area_a = max(0, ax2 - ax1) * max(0, ay2 - ay1)
    area_b = max(0, bx2 - bx1) * max(0, by2 - by1)

    union = area_a + area_b - intersection

    if union == 0:
        return 0.0

    return intersection / union


def evaluate_image(item, iou_threshold=0.50):

    image_path = item["image"]

    # Run YOLO
    results = model.predict(
        source=image_path,
        imgsz=640,
        conf=0.01,
        iou=0.50,
        verbose=False
    )

    result = results[0]

    predictions = []

    if result.boxes is not None:

        for box in result.boxes:

            xyxy = box.xyxy[0].cpu().numpy()

            # Standard YOLO model may detect several classes.
            # For this synthetic test we evaluate all predictions
            # as candidate detections.
            predictions.append(xyxy)

    ground_truth = [
        [
            box["x1"],
            box["y1"],
            box["x2"],
            box["y2"]
        ]
        for box in item["annotations"]
    ]

    matched_gt = set()
    true_positive = 0
    false_positive = 0

    # Match each prediction to the best unused ground-truth box
    for prediction in predictions:

        best_iou = 0
        best_gt = None

        for gt_index, gt in enumerate(ground_truth):

            if gt_index in matched_gt:
                continue

            current_iou = calculate_iou(
                prediction,
                gt
            )

            if current_iou > best_iou:
                best_iou = current_iou
                best_gt = gt_index

        if best_iou >= iou_threshold:
            true_positive += 1
            matched_gt.add(best_gt)
        else:
            false_positive += 1

    false_negative = len(ground_truth) - true_positive

    return {
        "image": os.path.basename(image_path),
        "ground_truth": len(ground_truth),
        "predictions": len(predictions),
        "TP": true_positive,
        "FP": false_positive,
        "FN": false_negative
    }


# Evaluate all five synthetic images
evaluation_results = []

for item in dataset:

    result = evaluate_image(item)

    evaluation_results.append(result)


evaluation_df = pd.DataFrame(evaluation_results)

print("=" * 70)
print("STORE VISION AI — YOLO DETECTION EVALUATION")
print("=" * 70)

print(evaluation_df.to_string(index=False))
```

```python

# Calculate overall detection metrics

TP = int(evaluation_df["TP"].sum())
FP = int(evaluation_df["FP"].sum())
FN = int(evaluation_df["FN"].sum())

precision = TP / (TP + FP) if (TP + FP) > 0 else 0
recall = TP / (TP + FN) if (TP + FN) > 0 else 0

ground_truth_total = int(
    evaluation_df["ground_truth"].sum()
)

prediction_total = int(
    evaluation_df["predictions"].sum()
)

count_error = prediction_total - ground_truth_total

print("=" * 50)
print("FINAL YOLO METRICS")
print("=" * 50)

print("Ground-truth objects :", ground_truth_total)
print("Predicted objects    :", prediction_total)
print("True positives       :", TP)
print("False positives      :", FP)
print("False negatives      :", FN)

print(f"\nPrecision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"Count error: {count_error}")

print("=" * 50)
```

```python

# ============================================================
# STORE VISION AI — PROCEDURAL VIRTUAL 3D STORE
# CELL 1: Environment + renderer + synthetic dataset generator
# ============================================================

!pip install -q opencv-python-headless pillow numpy pyyaml

import os
import math
import random
import json
import shutil
from pathlib import Path

import cv2
import numpy as np
import yaml
from PIL import Image, ImageDraw, ImageFont

# ------------------------------------------------------------
# 1. Reproducibility
# ------------------------------------------------------------

SEED = 2026
random.seed(SEED)
np.random.seed(SEED)

ROOT = Path("/content/store_vision_virtual3d")

if ROOT.exists():
    shutil.rmtree(ROOT)

DIRS = [
    ROOT / "dataset" / "images" / "train",
    ROOT / "dataset" / "images" / "val",
    ROOT / "dataset" / "images" / "test",
    ROOT / "dataset" / "labels" / "train",
    ROOT / "dataset" / "labels" / "val",
    ROOT / "dataset" / "labels" / "test",
    ROOT / "video",
    ROOT / "metadata",
]

for d in DIRS:
    d.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2. Canvas
# ------------------------------------------------------------

W, H = 960, 540

# ------------------------------------------------------------
# 3. Utility functions
# ------------------------------------------------------------

def clamp(v, lo, hi):
    return max(lo, min(hi, v))


def lerp(a, b, t):
    return a + (b - a) * t


def project_point(x, y, z, camera_x=0.0, camera_y=1.8,
                  camera_z=-8.0, focal=650):
    """
    Lightweight perspective projection.

    x = left/right
    y = vertical
    z = depth
    """

    dz = z - camera_z

    if dz < 0.5:
        dz = 0.5

    sx = W / 2 + (x - camera_x) * focal / dz

    sy = H * 0.58 - (y - camera_y) * focal / dz

    return int(sx), int(sy)


def polygon(draw, points, fill, outline=None, width=1):
    draw.polygon(points, fill=fill)

    if outline:
        draw.line(
            points + [points[0]],
            fill=outline,
            width=width
        )


def random_product_color():
    palettes = [
        (220, 55, 55),
        (50, 110, 220),
        (55, 170, 85),
        (235, 165, 40),
        (150, 75, 190),
        (35, 170, 175),
        (210, 105, 55),
        (90, 90, 100),
        (200, 70, 125),
    ]

    return random.choice(palettes)


# ------------------------------------------------------------
# 4. Draw 3D cuboid product
# ------------------------------------------------------------

def draw_product(draw, x, y, z, pw, ph, pd, color,
                 camera_x=0.0):

    # Front-bottom-left/right
    p1 = project_point(
        x - pw/2, y, z,
        camera_x=camera_x
    )

    p2 = project_point(
        x + pw/2, y, z,
        camera_x=camera_x
    )

    # Front-top-left/right
    p3 = project_point(
        x - pw/2, y + ph, z,
        camera_x=camera_x
    )

    p4 = project_point(
        x + pw/2, y + ph, z,
        camera_x=camera_x
    )

    # Back points
    p5 = project_point(
        x - pw/2, y, z + pd,
        camera_x=camera_x
    )

    p6 = project_point(
        x + pw/2, y, z + pd,
        camera_x=camera_x
    )

    p7 = project_point(
        x - pw/2, y + ph, z + pd,
        camera_x=camera_x
    )

    p8 = project_point(
        x + pw/2, y + ph, z + pd,
        camera_x=camera_x
    )

    # Slightly different shades for 3D appearance
    front = color

    side = tuple(max(0, int(c * 0.72)) for c in color)
    top = tuple(min(255, int(c * 1.20)) for c in color)

    # Back/side
    polygon(
        draw,
        [p2, p6, p8, p4],
        side,
        outline=(25, 25, 25),
        width=2
    )

    # Top
    polygon(
        draw,
        [p3, p4, p8, p7],
        top,
        outline=(25, 25, 25),
        width=2
    )

    # Front
    polygon(
        draw,
        [p1, p2, p4, p3],
        front,
        outline=(20, 20, 20),
        width=2
    )

    # Product label
    cx = int((p3[0] + p4[0]) / 2)
    cy = int((p3[1] + p1[1]) / 2)

    label_w = max(4, abs(p4[0] - p3[0]) // 3)
    label_h = max(3, abs(p1[1] - p3[1]) // 4)

    draw.rectangle(
        [
            cx - label_w,
            cy - label_h,
            cx + label_w,
            cy + label_h
        ],
        fill=(245, 245, 235),
        outline=(40, 40, 40),
        width=1
    )

    # Exact projected bounding box
    pts = [p1, p2, p3, p4, p5, p6, p7, p8]

    xs = [p[0] for p in pts]
    ys = [p[1] for p in pts]

    xmin = clamp(min(xs), 0, W - 1)
    xmax = clamp(max(xs), 0, W - 1)
    ymin = clamp(min(ys), 0, H - 1)
    ymax = clamp(max(ys), 0, H - 1)

    return [xmin, ymin, xmax, ymax]


# ------------------------------------------------------------
# 5. Draw virtual customer
# ------------------------------------------------------------

def draw_customer(draw, x, y, z, scale=1.0, camera_x=0.0):

    # Head
    head_center = project_point(
        x,
        y + 1.55 * scale,
        z,
        camera_x=camera_x
    )

    head_r = max(
        4,
        int(13 * scale * 650 /
            max(1, z + 8))
    )

    draw.ellipse(
        [
            head_center[0] - head_r,
            head_center[1] - head_r,
            head_center[0] + head_r,
            head_center[1] + head_r
        ],
        fill=(225, 185, 150),
        outline=(30, 30, 30)
    )

    # Body
    left = project_point(
        x - 0.28 * scale,
        y,
        z,
        camera_x=camera_x
    )

    right = project_point(
        x + 0.28 * scale,
        y,
        z,
        camera_x=camera_x
    )

    top = project_point(
        x,
        y + 1.35 * scale,
        z,
        camera_x=camera_x
    )

    polygon(
        draw,
        [
            left,
            right,
            (
                right[0],
                right[1] + int(65 * scale)
            ),
            (
                left[0],
                left[1] + int(65 * scale)
            )
        ],
        fill=(65, 75, 95),
        outline=(25, 25, 25)
    )


# ------------------------------------------------------------
# 6. Virtual store scene
# ------------------------------------------------------------

def create_store_scene(scene_seed,
                       camera_x=0.0,
                       customer_offset=0.0,
                       num_products=None):

    random.seed(scene_seed)

    img = Image.new(
        "RGB",
        (W, H),
        (225, 225, 225)
    )

    draw = ImageDraw.Draw(img)

    # --------------------------------------------------------
    # Back wall
    # --------------------------------------------------------

    draw.rectangle(
        [0, 0, W, int(H * 0.60)],
        fill=(205, 210, 218)
    )

    # Ceiling
    draw.polygon(
        [
            (0, 0),
            (W, 0),
            (W, 80),
            (0, 80)
        ],
        fill=(235, 237, 240)
    )

    # Floor
    draw.polygon(
        [
            (0, int(H * 0.58)),
            (W, int(H * 0.58)),
            (W, H),
            (0, H)
        ],
        fill=(165, 168, 172)
    )

    # Floor perspective lines
    vanishing = (W // 2, int(H * 0.58))

    for x in range(-600, W + 601, 120):

        draw.line(
            [
                vanishing,
                (x, H)
            ],
            fill=(135, 138, 142),
            width=2
        )

    for yy in [420, 470, 515]:

        draw.line(
            [
                (0, yy),
                (W, yy)
            ],
            fill=(145, 148, 152),
            width=2
        )

    # --------------------------------------------------------
    # Store sign
    # --------------------------------------------------------

    draw.rounded_rectangle(
        [330, 25, 630, 80],
        radius=12,
        fill=(45, 50, 60),
        outline=(15, 15, 15),
        width=3
    )

    draw.text(
        (400, 42),
        "VIRTUAL MART",
        fill=(245, 245, 245)
    )

    # --------------------------------------------------------
    # Shelves
    # --------------------------------------------------------

    shelf_x_positions = [-3.4, 0.0, 3.4]

    shelf_levels = [0.7, 1.8, 2.9]

    products = []

    for sx in shelf_x_positions:

        # Shelf vertical supports
        for side in [-1, 1]:

            px = sx + side * 1.25

            a = project_point(
                px, 0.25, 3.0,
                camera_x=camera_x
            )

            b = project_point(
                px, 3.8, 3.0,
                camera_x=camera_x
            )

            draw.line(
                [a, b],
                fill=(55, 60, 65),
                width=8
            )

        # Shelf boards
        for level in shelf_levels:

            left = project_point(
                sx - 1.35,
                level,
                2.8,
                camera_x=camera_x
            )

            right = project_point(
                sx + 1.35,
                level,
                2.8,
                camera_x=camera_x
            )

            draw.line(
                [left, right],
                fill=(65, 70, 75),
                width=12
            )

        # Products
        for level in shelf_levels:

            count = random.randint(5, 8)

            for i in range(count):

                px = (
                    sx - 1.05
                    + i * (2.1 / max(1, count - 1))
                    + random.uniform(-0.08, 0.08)
                )

                ph = random.uniform(0.45, 0.75)
                pw = random.uniform(0.25, 0.42)
                pd = random.uniform(0.18, 0.35)

                py = level

                pz = 2.55 + random.uniform(-0.05, 0.10)

                bbox = draw_product(
                    draw,
                    px,
                    py,
                    pz,
                    pw,
                    ph,
                    pd,
                    random_product_color(),
                    camera_x=camera_x
                )

                xmin, ymin, xmax, ymax = bbox

                # Ignore objects outside the useful frame
                if (
                    xmax - xmin >= 4
                    and ymax - ymin >= 6
                    and xmax > 0
                    and xmin < W
                    and ymax > 0
                    and ymin < H
                ):
                    products.append({
                        "bbox": bbox,
                        "class": 0
                    })

    # --------------------------------------------------------
    # Checkout counter
    # --------------------------------------------------------

    draw.polygon(
        [
            (690, 335),
            (900, 365),
            (900, 470),
            (690, 440)
        ],
        fill=(95, 70, 55),
        outline=(35, 35, 35)
    )

    draw.polygon(
        [
            (690, 335),
            (810, 305),
            (900, 335),
            (780, 365)
        ],
        fill=(130, 95, 70),
        outline=(35, 35, 35)
    )

    # Checkout scanner
    draw.rectangle(
        [735, 330, 770, 350],
        fill=(40, 40, 45),
        outline=(10, 10, 10)
    )

    # --------------------------------------------------------
    # Customers
    # --------------------------------------------------------

    customers = []

    customer_x = -1.0 + customer_offset

    if -5 < customer_x < 5:

        draw_customer(
            draw,
            customer_x,
            0.2,
            1.1,
            scale=1.2,
            camera_x=camera_x
        )

        customers.append({
            "x": customer_x,
            "z": 1.1
        })

    # Second customer occasionally
    if scene_seed % 3 == 0:

        draw_customer(
            draw,
            2.2,
            0.15,
            0.4,
            scale=0.9,
            camera_x=camera_x
        )

        customers.append({
            "x": 2.2,
            "z": 0.4
        })

    # --------------------------------------------------------
    # Return image + exact ground truth
    # --------------------------------------------------------

    return img, products, customers


# ------------------------------------------------------------
# 7. YOLO label conversion
# ------------------------------------------------------------

def bbox_to_yolo(bbox):

    xmin, ymin, xmax, ymax = bbox

    xc = ((xmin + xmax) / 2) / W
    yc = ((ymin + ymax) / 2) / H

    bw = (xmax - xmin) / W
    bh = (ymax - ymin) / H

    return (
        0,
        xc,
        yc,
        bw,
        bh
    )


def save_scene(image,
               products,
               image_path,
               label_path):

    image.save(image_path)

    with open(label_path, "w") as f:

        for product in products:

            row = bbox_to_yolo(
                product["bbox"]
            )

            f.write(
                f"{row[0]} "
                f"{row[1]:.6f} "
                f"{row[2]:.6f} "
                f"{row[3]:.6f} "
                f"{row[4]:.6f}\n"
            )


# ------------------------------------------------------------
# 8. Generate dataset
# ------------------------------------------------------------

TOTAL_IMAGES = 45

records = []

indices = list(range(TOTAL_IMAGES))
random.shuffle(indices)

train_ids = set(indices[:32])
val_ids = set(indices[32:39])
test_ids = set(indices[39:])

for i in range(TOTAL_IMAGES):

    scene_seed = SEED + i * 97

    camera_x = random.uniform(-0.8, 0.8)

    image, products, customers = create_store_scene(
        scene_seed,
        camera_x=camera_x,
        customer_offset=random.uniform(-1.5, 1.5)
    )

    if i in train_ids:
        split = "train"
    elif i in val_ids:
        split = "val"
    else:
        split = "test"

    image_name = f"virtual_store_{i:04d}.jpg"

    image_path = (
        ROOT /
        "dataset" /
        "images" /
        split /
        image_name
    )

    label_path = (
        ROOT /
        "dataset" /
        "labels" /
        split /
        image_name.replace(
            ".jpg",
            ".txt"
        )
    )

    save_scene(
        image,
        products,
        image_path,
        label_path
    )

    records.append({
        "image": str(image_path),
        "split": split,
        "products": len(products),
        "customers": len(customers),
        "camera_x": camera_x
    })


# ------------------------------------------------------------
# 9. Create YOLO data.yaml
# ------------------------------------------------------------

data_yaml = {
    "path": str(ROOT / "dataset"),
    "train": "images/train",
    "val": "images/val",
    "test": "images/test",
    "names": {
        0: "product"
    }
}

with open(
    ROOT / "dataset" / "data.yaml",
    "w"
) as f:

    yaml.safe_dump(
        data_yaml,
        f,
        sort_keys=False
    )


# ------------------------------------------------------------
# 10. Save metadata
# ------------------------------------------------------------

with open(
    ROOT / "metadata" / "dataset.json",
    "w"
) as f:

    json.dump(
        records,
        f,
        indent=2
    )


# ------------------------------------------------------------
# 11. Summary
# ------------------------------------------------------------

total_objects = sum(
    r["products"]
    for r in records
)

print("=" * 60)
print("STORE VISION AI — VIRTUAL 3D DATASET")
print("=" * 60)
print(f"Images generated : {TOTAL_IMAGES}")
print(f"Training images  : {len(train_ids)}")
print(f"Validation images: {len(val_ids)}")
print(f"Test images      : {len(test_ids)}")
print(f"Product labels   : {total_objects}")
print()
print("Dataset:")
print(ROOT / "dataset")
print()
print("YOLO YAML:")
print(ROOT / "dataset" / "data.yaml")
print()
print("✅ Procedural virtual store created")
print("✅ Products generated by code")
print("✅ Automatic YOLO bounding boxes")
print("✅ Train/validation/test split")
print("✅ No real-world images used")
print("✅ No external dataset used")
print("✅ No third-party media assets used")
print("=" * 60)
```

```python

# ============================================================
# STORE VISION AI
# CELL 2: Dataset verification + virtual store video
# ============================================================

from pathlib import Path
import cv2
import random
import json
import numpy as np
from PIL import Image, ImageDraw

ROOT = Path("/content/store_vision_virtual3d")
DATASET = ROOT / "dataset"

W, H = 960, 540

# ------------------------------------------------------------
# 1. Verify image/label pairs
# ------------------------------------------------------------

print("=" * 60)
print("DATASET VERIFICATION")
print("=" * 60)

splits = ["train", "val", "test"]

total_images = 0
total_labels = 0

for split in splits:

    image_dir = DATASET / "images" / split
    label_dir = DATASET / "labels" / split

    images = sorted(image_dir.glob("*.jpg"))

    missing_labels = []
    invalid_labels = []
    object_count = 0

    for image_path in images:

        label_path = label_dir / (
            image_path.stem + ".txt"
        )

        if not label_path.exists():
            missing_labels.append(image_path.name)
            continue

        with open(label_path) as f:
            lines = [
                x.strip()
                for x in f.readlines()
                if x.strip()
            ]

        for line in lines:

            parts = line.split()

            if len(parts) != 5:
                invalid_labels.append(
                    (image_path.name, line)
                )
                continue

            cls, xc, yc, bw, bh = map(
                float,
                parts
            )

            # YOLO normalized coordinates
            if not (
                0 <= xc <= 1 and
                0 <= yc <= 1 and
                0 < bw <= 1 and
                0 < bh <= 1
            ):
                invalid_labels.append(
                    (image_path.name, line)
                )

            if int(cls) != 0:
                invalid_labels.append(
                    (image_path.name, line)
                )

            object_count += 1

    total_images += len(images)
    total_labels += object_count

    print()
    print(f"{split.upper()}")
    print("-" * 40)
    print(f"Images          : {len(images)}")
    print(f"Product labels  : {object_count}")
    print(f"Missing labels  : {len(missing_labels)}")
    print(f"Invalid labels  : {len(invalid_labels)}")

    if missing_labels:
        print("⚠️ Missing:", missing_labels[:5])

    if invalid_labels:
        print("⚠️ Invalid:", invalid_labels[:3])


print()
print("=" * 60)
print(f"TOTAL IMAGES  : {total_images}")
print(f"TOTAL LABELS  : {total_labels}")
print("=" * 60)


# ------------------------------------------------------------
# 2. Create visual verification images
# ------------------------------------------------------------

def draw_yolo_boxes(image_path, label_path):

    img = cv2.imread(str(image_path))

    if img is None:
        return None

    h, w = img.shape[:2]

    with open(label_path) as f:
        lines = [
            x.strip()
            for x in f.readlines()
            if x.strip()
        ]

    for line in lines:

        cls, xc, yc, bw, bh = map(
            float,
            line.split()
        )

        x1 = int((xc - bw / 2) * w)
        y1 = int((yc - bh / 2) * h)
        x2 = int((xc + bw / 2) * w)
        y2 = int((yc + bh / 2) * h)

        x1 = max(0, min(w - 1, x1))
        y1 = max(0, min(h - 1, y1))
        x2 = max(0, min(w - 1, x2))
        y2 = max(0, min(h - 1, y2))

        cv2.rectangle(
            img,
            (x1, y1),
            (x2, y2),
            (0, 255, 0),
            2
        )

        cv2.putText(
            img,
            "product",
            (x1, max(15, y1 - 4)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.45,
            (0, 255, 0),
            1,
            cv2.LINE_AA
        )

    return img


verify_dir = ROOT / "verification"
verify_dir.mkdir(exist_ok=True)

random.seed(2026)

test_images = sorted(
    (DATASET / "images" / "test").glob("*.jpg")
)

for i, image_path in enumerate(test_images):

    label_path = (
        DATASET /
        "labels" /
        "test" /
        f"{image_path.stem}.txt"
    )

    result = draw_yolo_boxes(
        image_path,
        label_path
    )

    if result is not None:

        output = (
            verify_dir /
            f"verified_{i:02d}.jpg"
        )

        cv2.imwrite(
            str(output),
            result
        )


# ------------------------------------------------------------
# 3. Generate virtual-store video
# ------------------------------------------------------------

print()
print("=" * 60)
print("GENERATING VIRTUAL STORE VIDEO")
print("=" * 60)

# Reuse the renderer from Cell 1
# create_store_scene(...)

VIDEO_DIR = ROOT / "video"
VIDEO_DIR.mkdir(exist_ok=True)

video_path = VIDEO_DIR / "virtual_store_walkthrough.mp4"

FPS = 8
FRAMES = 64

fourcc = cv2.VideoWriter_fourcc(
    *"mp4v"
)

writer = cv2.VideoWriter(
    str(video_path),
    fourcc,
    FPS,
    (W, H)
)

video_metadata = []

for frame_id in range(FRAMES):

    # Smooth camera movement
    camera_x = (
        math.sin(frame_id / 12.0)
        * 0.75
    )

    # Customer moves through store
    customer_offset = (
        -2.5
        + frame_id * 0.075
    )

    image, products, customers = create_store_scene(
        scene_seed=5000 + frame_id,
        camera_x=camera_x,
        customer_offset=customer_offset
    )

    frame = np.array(image)

    frame = cv2.cvtColor(
        frame,
        cv2.COLOR_RGB2BGR
    )

    writer.write(frame)

    video_metadata.append({
        "frame": frame_id,
        "products": len(products),
        "customers": len(customers),
        "camera_x": camera_x,
        "customer_offset": customer_offset
    })


writer.release()


# ------------------------------------------------------------
# 4. Save video metadata
# ------------------------------------------------------------

metadata_path = (
    ROOT /
    "metadata" /
    "video_metadata.json"
)

with open(metadata_path, "w") as f:

    json.dump(
        video_metadata,
        f,
        indent=2
    )


# ------------------------------------------------------------
# 5. Extract several video frames for testing
# ------------------------------------------------------------

frame_dir = (
    VIDEO_DIR /
    "frames"
)

frame_dir.mkdir(exist_ok=True)

cap = cv2.VideoCapture(
    str(video_path)
)

frame_number = 0
saved_frames = 0

while True:

    ok, frame = cap.read()

    if not ok:
        break

    if frame_number % 8 == 0:

        frame_path = (
            frame_dir /
            f"frame_{frame_number:04d}.jpg"
        )

        cv2.imwrite(
            str(frame_path),
            frame
        )

        saved_frames += 1

    frame_number += 1

cap.release()


print()
print("=" * 60)
print("CELL 2 COMPLETE")
print("=" * 60)
print(f"Video frames generated : {FRAMES}")
print(f"Video FPS              : {FPS}")
print(f"Video duration         : {FRAMES / FPS:.1f} seconds")
print(f"Video                  : {video_path}")
print(f"Extracted test frames  : {saved_frames}")
print(f"Verification folder    : {verify_dir}")
print()
print("✅ YOLO labels checked")
print("✅ Bounding boxes converted and verified")
print("✅ Virtual store video generated")
print("✅ Moving virtual customer generated")
print("✅ Moving camera generated")
print("✅ Video metadata saved")
print("✅ No real-world media used")
print("=" * 60)
```

```python

# ============================================================
# STORE VISION AI — SYNTHETIC TEST ENVIRONMENT
# ============================================================

import os
import json
import random
from PIL import Image, ImageDraw

BASE = "/content/store_vision_synthetic"
IMG_DIR = os.path.join(BASE, "images")
ANN_DIR = os.path.join(BASE, "annotations")

os.makedirs(IMG_DIR, exist_ok=True)
os.makedirs(ANN_DIR, exist_ok=True)

random.seed(42)

# Synthetic retail classes
CLASSES = {
    0: "person",
    1: "bottle",
    2: "box",
    3: "book",
    4: "cup",
    5: "bag"
}

# Generate a small lightweight dataset
NUM_IMAGES = 12
IMG_W = 640
IMG_H = 480

print("=" * 60)
print("STORE VISION AI — SYNTHETIC DATA GENERATION")
print("=" * 60)

for i in range(NUM_IMAGES):

    img = Image.new("RGB", (IMG_W, IMG_H), "white")
    draw = ImageDraw.Draw(img)

    objects = []

    # Random number of objects
    n = random.randint(3, 8)

    for _ in range(n):

        class_id = random.choice(list(CLASSES.keys()))

        w = random.randint(40, 100)
        h = random.randint(40, 120)

        x1 = random.randint(10, IMG_W - w - 10)
        y1 = random.randint(10, IMG_H - h - 10)

        x2 = x1 + w
        y2 = y1 + h

        # Draw synthetic object
        draw.rectangle(
            [x1, y1, x2, y2],
            outline="black",
            width=3
        )

        draw.text(
            (x1 + 3, y1 + 3),
            CLASSES[class_id],
            fill="black"
        )

        objects.append({
            "class_id": class_id,
            "class_name": CLASSES[class_id],
            "bbox": [x1, y1, x2, y2]
        })

    image_path = os.path.join(
        IMG_DIR,
        f"synthetic_store_{i:03d}.jpg"
    )

    annotation_path = os.path.join(
        ANN_DIR,
        f"synthetic_store_{i:03d}.json"
    )

    img.save(image_path, quality=90)

    with open(annotation_path, "w") as f:
        json.dump(
            {
                "image": image_path,
                "width": IMG_W,
                "height": IMG_H,
                "objects": objects
            },
            f,
            indent=2
        )

print(f"✅ Generated images: {NUM_IMAGES}")
print(f"✅ Image directory: {IMG_DIR}")
print(f"✅ Annotation directory: {ANN_DIR}")

print("\nClasses:")
for k, v in CLASSES.items():
    print(f"  {k}: {v}")

print("=" * 60)
print("✅ SYNTHETIC DATASET READY")
print("=" * 60)
```

```python

# ============================================================
# STORE VISION AI — BILLING / OBSERVATION LOGIC TEST
# ============================================================

import os
import json
from collections import Counter

BASE = "/content/store_vision_synthetic"
ANN_DIR = os.path.join(BASE, "annotations")

print("=" * 60)
print("STORE VISION AI — BILLING COMPARISON TEST")
print("=" * 60)

# ------------------------------------------------------------
# Load all observed objects
# ------------------------------------------------------------

observed = Counter()

for filename in sorted(os.listdir(ANN_DIR)):

    if not filename.endswith(".json"):
        continue

    with open(os.path.join(ANN_DIR, filename), "r") as f:
        data = json.load(f)

    for obj in data["objects"]:
        observed[obj["class_name"]] += 1

print("OBSERVED BY VISION")
print("-" * 40)

for item, quantity in sorted(observed.items()):
    print(f"{item:10s}: {quantity}")

# ------------------------------------------------------------
# Create synthetic billing record
#
# Intentionally introduce discrepancies so the
# alert logic can be tested.
# ------------------------------------------------------------

billed = Counter(observed)

# One item observed but NOT billed
billed["bottle"] -= 2

# One item billed but NOT observed
billed["book"] += 1

# Remove zero/negative quantities
billed = Counter({
    item: qty
    for item, qty in billed.items()
    if qty > 0
})

print("\nBILLED")
print("-" * 40)

for item, quantity in sorted(billed.items()):
    print(f"{item:10s}: {quantity}")

# ------------------------------------------------------------
# Compare vision observations against billing
# ------------------------------------------------------------

alerts = []
missing_from_vision = []

all_items = set(observed) | set(billed)

for item in sorted(all_items):

    observed_qty = observed.get(item, 0)
    billed_qty = billed.get(item, 0)

    if observed_qty > billed_qty:

        difference = observed_qty - billed_qty

        alerts.append({
            "item": item,
            "observed": observed_qty,
            "billed": billed_qty,
            "unbilled_quantity": difference,
            "type": "OBSERVED_NOT_BILLED"
        })

    elif billed_qty > observed_qty:

        difference = billed_qty - observed_qty

        missing_from_vision.append({
            "item": item,
            "observed": observed_qty,
            "billed": billed_qty,
            "difference": difference,
            "type": "BILLED_NOT_OBSERVED"
        })

# ------------------------------------------------------------
# Results
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DISCREPANCY RESULTS")
print("=" * 60)

print("\n🚨 OBSERVED BUT NOT BILLED")

if alerts:
    for alert in alerts:
        print(
            f"  {alert['item']}: "
            f"observed={alert['observed']}, "
            f"billed={alert['billed']}, "
            f"unbilled={alert['unbilled_quantity']}"
        )
else:
    print("  None")

print("\n⚠️ BILLED BUT NOT OBSERVED")

if missing_from_vision:
    for item in missing_from_vision:
        print(
            f"  {item['item']}: "
            f"observed={item['observed']}, "
            f"billed={item['billed']}, "
            f"difference={item['difference']}"
        )
else:
    print("  None")

# ------------------------------------------------------------
# Assertions
# ------------------------------------------------------------

assert len(alerts) > 0, \
    "Expected an observed-but-unbilled discrepancy"

assert any(
    a["item"] == "bottle" and
    a["unbilled_quantity"] == 2
    for a in alerts
), "Bottle discrepancy was not detected"

assert any(
    a["item"] == "book"
    for a in missing_from_vision
), "Billed-but-unobserved book discrepancy was not detected"

print("\n" + "=" * 60)
print("✅ OBSERVATION/BILLING COMPARISON PASSED")
print("✅ UNBILLED ITEM DETECTION PASSED")
print("✅ BILLING DISCREPANCY DETECTION PASSED")
print("=" * 60)
```

```python

# ============================================================
# STORE VISION AI — CUSTOMER JOURNEY / ITEM ASSOCIATION TEST
# ============================================================

from collections import defaultdict

print("=" * 60)
print("STORE VISION AI — CUSTOMER JOURNEY TEST")
print("=" * 60)

# Synthetic customer events.
# Each customer has a simple entry -> observation -> checkout path.

events = [
    {"frame": 1,  "customer": "C001", "event": "ENTRY"},
    {"frame": 8,  "customer": "C001", "event": "ITEM_OBSERVED",
     "item": "bottle"},
    {"frame": 15, "customer": "C001", "event": "ITEM_OBSERVED",
     "item": "book"},
    {"frame": 25, "customer": "C001", "event": "CHECKOUT"},
    {"frame": 30, "customer": "C001", "event": "EXIT"},

    {"frame": 2,  "customer": "C002", "event": "ENTRY"},
    {"frame": 10, "customer": "C002", "event": "ITEM_OBSERVED",
     "item": "bag"},
    {"frame": 22, "customer": "C002", "event": "CHECKOUT"},
    {"frame": 28, "customer": "C002", "event": "EXIT"},

    {"frame": 3,  "customer": "C003", "event": "ENTRY"},
    {"frame": 12, "customer": "C003", "event": "ITEM_OBSERVED",
     "item": "bottle"},
    {"frame": 20, "customer": "C003", "event": "CHECKOUT"},
    {"frame": 27, "customer": "C003", "event": "EXIT"},
]

# ------------------------------------------------------------
# Build customer journeys
# ------------------------------------------------------------

journeys = defaultdict(lambda: {
    "entry": None,
    "checkout": None,
    "exit": None,
    "items": []
})

for event in events:

    customer = event["customer"]
    event_type = event["event"]

    if event_type == "ENTRY":
        journeys[customer]["entry"] = event["frame"]

    elif event_type == "CHECKOUT":
        journeys[customer]["checkout"] = event["frame"]

    elif event_type == "EXIT":
        journeys[customer]["exit"] = event["frame"]

    elif event_type == "ITEM_OBSERVED":
        journeys[customer]["items"].append({
            "frame": event["frame"],
            "item": event["item"]
        })

# ------------------------------------------------------------
# Validate customer journeys
# ------------------------------------------------------------

print("\nCUSTOMER JOURNEYS")
print("-" * 60)

for customer, journey in sorted(journeys.items()):

    print(f"\n{customer}")
    print(f"  Entry    : frame {journey['entry']}")
    print(f"  Items    : {[x['item'] for x in journey['items']]}")
    print(f"  Checkout : frame {journey['checkout']}")
    print(f"  Exit     : frame {journey['exit']}")

    assert journey["entry"] is not None
    assert journey["exit"] is not None

    assert journey["entry"] < journey["exit"]

    for item in journey["items"]:
        assert journey["entry"] < item["frame"] < journey["exit"]

# ------------------------------------------------------------
# Calculate visit duration
# ------------------------------------------------------------

print("\nVISIT ANALYTICS")
print("-" * 60)

for customer, journey in sorted(journeys.items()):

    duration = journey["exit"] - journey["entry"]

    print(
        f"{customer}: "
        f"visit_duration={duration} frames, "
        f"items_observed={len(journey['items'])}"
    )

    assert duration > 0

# ------------------------------------------------------------
# Test item/customer association
# ------------------------------------------------------------

association = []

for customer, journey in journeys.items():

    for item in journey["items"]:

        association.append({
            "customer": customer,
            "item": item["item"],
            "frame": item["frame"]
        })

print("\nITEM-CUSTOMER ASSOCIATION")
print("-" * 60)

for record in association:
    print(
        f"  {record['customer']} -> "
        f"{record['item']} "
        f"(frame {record['frame']})"
    )

assert len(association) == 4

# Verify expected associations
assert any(
    x["customer"] == "C001" and x["item"] == "bottle"
    for x in association
)

assert any(
    x["customer"] == "C001" and x["item"] == "book"
    for x in association
)

assert any(
    x["customer"] == "C002" and x["item"] == "bag"
    for x in association
)

assert any(
    x["customer"] == "C003" and x["item"] == "bottle"
    for x in association
)

print("\n" + "=" * 60)
print("✅ ENTRY DETECTION PASSED")
print("✅ EXIT DETECTION PASSED")
print("✅ CUSTOMER JOURNEY PASSED")
print("✅ VISIT DURATION PASSED")
print("✅ ITEM-CUSTOMER ASSOCIATION PASSED")
print("=" * 60)
```

```python

# ============================================================
# STORE VISION AI — THEFT / UNPAID ITEM TEST
# Synthetic data only — no images or large datasets required
# ============================================================

from collections import Counter

print("=" * 60)
print("STORE VISION AI — THEFT / UNPAID ITEM TEST")
print("=" * 60)

# ------------------------------------------------------------
# SYNTHETIC CUSTOMER OBSERVATIONS
# ------------------------------------------------------------

observed_items = {
    "C001": ["bottle", "book"],
    "C002": ["bag"],
    "C003": ["bottle", "book", "bag"],
}

# ------------------------------------------------------------
# SYNTHETIC BILLING DATA
# ------------------------------------------------------------

billed_items = {
    "C001": ["bottle", "book"],       # Everything billed
    "C002": ["bag"],                  # Everything billed
    "C003": ["bottle", "bag"],        # BOOK NOT BILLED
}

# ------------------------------------------------------------
# DETECTION FUNCTION
# ------------------------------------------------------------

def compare_customer_billing(observed, billed):
    observed_count = Counter(observed)
    billed_count = Counter(billed)

    unpaid = []

    for item, count in observed_count.items():
        difference = count - billed_count.get(item, 0)

        if difference > 0:
            unpaid.extend([item] * difference)

    return unpaid


# ------------------------------------------------------------
# RUN BILLING COMPARISON
# ------------------------------------------------------------

alerts = []

print("\nCUSTOMER BILLING ANALYSIS")
print("-" * 60)

for customer_id in observed_items:

    observed = observed_items[customer_id]
    billed = billed_items.get(customer_id, [])

    unpaid = compare_customer_billing(observed, billed)

    print(f"\n{customer_id}")
    print(f"  Observed : {observed}")
    print(f"  Billed   : {billed}")

    if unpaid:
        alert = {
            "customer_id": customer_id,
            "items": unpaid,
            "alert_type": "SUSPECTED_UNPAID_ITEM",
        }

        alerts.append(alert)

        print(f"  ⚠️ ALERT  : suspected unpaid item(s) = {unpaid}")

    else:
        print("  ✅ Status : all observed items accounted for")


# ------------------------------------------------------------
# ALERT SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("SECURITY ALERT SUMMARY")
print("-" * 60)

print(f"Total customers checked : {len(observed_items)}")
print(f"Alerts generated        : {len(alerts)}")

for alert in alerts:
    print(
        f"  {alert['customer_id']} -> "
        f"{alert['items']} -> "
        f"{alert['alert_type']}"
    )


# ------------------------------------------------------------
# AUTOMATED ASSERTIONS
# ------------------------------------------------------------

assert len(alerts) == 1, "Expected exactly one security alert"

assert alerts[0]["customer_id"] == "C003", \
    "Expected C003 to trigger the alert"

assert alerts[0]["items"] == ["book"], \
    "Expected the unpaid item to be book"

print("\n" + "=" * 60)
print("TEST RESULTS")
print("=" * 60)

print("✅ BILLING COMPARISON PASSED")
print("✅ UNPAID ITEM DETECTION PASSED")
print("✅ CUSTOMER-SPECIFIC ALERT PASSED")
print("✅ FALSE ALERT PREVENTION PASSED")
print("✅ SECURITY ALERT GENERATION PASSED")

print("=" * 60)
print("🎯 THEFT / UNPAID ITEM TEST PASSED")
print("=" * 60)
```

```python

# ============================================================
# STORE VISION AI — IMAGE ANALYSIS TEST
# Synthetic image-analysis validation
# ============================================================

import numpy as np
import cv2
from collections import Counter

print("=" * 60)
print("STORE VISION AI — IMAGE ANALYSIS TEST")
print("=" * 60)


# ------------------------------------------------------------
# 1. CREATE SYNTHETIC STORE IMAGE
# ------------------------------------------------------------

IMAGE_WIDTH = 640
IMAGE_HEIGHT = 480

image = np.zeros(
    (IMAGE_HEIGHT, IMAGE_WIDTH, 3),
    dtype=np.uint8
)

# Create simple synthetic product regions.
# These are intentionally simple shapes so the test
# does not require downloading a model or dataset.

objects = [
    {
        "class": "bottle",
        "x1": 80,
        "y1": 100,
        "x2": 150,
        "y2": 300,
        "confidence": 0.94,
    },
    {
        "class": "book",
        "x1": 200,
        "y1": 120,
        "x2": 330,
        "y2": 280,
        "confidence": 0.91,
    },
    {
        "class": "bag",
        "x1": 380,
        "y1": 150,
        "x2": 540,
        "y2": 320,
        "confidence": 0.89,
    },
]


# ------------------------------------------------------------
# 2. DRAW SYNTHETIC OBJECTS
# ------------------------------------------------------------

for obj in objects:

    cv2.rectangle(
        image,
        (obj["x1"], obj["y1"]),
        (obj["x2"], obj["y2"]),
        (255, 255, 255),
        -1,
    )


# ------------------------------------------------------------
# 3. IMAGE VALIDATION
# ------------------------------------------------------------

assert image is not None, "Image creation failed"

assert image.shape == (
    IMAGE_HEIGHT,
    IMAGE_WIDTH,
    3,
), "Unexpected image dimensions"

assert image.dtype == np.uint8, \
    "Image must use uint8 format"

print("\nIMAGE VALIDATION")
print("-" * 60)

print(f"Image width  : {image.shape[1]}")
print(f"Image height : {image.shape[0]}")
print(f"Channels     : {image.shape[2]}")
print(f"Data type    : {image.dtype}")

print("✅ Synthetic image created")
print("✅ Image dimensions valid")
print("✅ Image format valid")


# ------------------------------------------------------------
# 4. SIMULATED VISION DETECTIONS
# ------------------------------------------------------------

# This represents the structured output that a detector
# such as YOLO would provide to the Store Vision AI pipeline.

detections = []

for obj in objects:

    detections.append({
        "class": obj["class"],
        "confidence": obj["confidence"],
        "bbox": [
            obj["x1"],
            obj["y1"],
            obj["x2"],
            obj["y2"],
        ],
    })


# ------------------------------------------------------------
# 5. CONFIDENCE FILTER
# ------------------------------------------------------------

CONFIDENCE_THRESHOLD = 0.50

filtered_detections = [
    detection
    for detection in detections
    if detection["confidence"] >= CONFIDENCE_THRESHOLD
]


# ------------------------------------------------------------
# 6. DETECTION VALIDATION
# ------------------------------------------------------------

print("\nDETECTION ANALYSIS")
print("-" * 60)

print(
    f"Raw detections      : "
    f"{len(detections)}"
)

print(
    f"Confidence threshold: "
    f"{CONFIDENCE_THRESHOLD}"
)

print(
    f"Valid detections     : "
    f"{len(filtered_detections)}"
)

for detection in filtered_detections:

    print(
        f"  {detection['class']:<8} "
        f"confidence={detection['confidence']:.2f} "
        f"bbox={detection['bbox']}"
    )


assert len(filtered_detections) == 3, \
    "Expected 3 valid detections"


# ------------------------------------------------------------
# 7. OBJECT COUNTING
# ------------------------------------------------------------

class_counts = Counter(
    detection["class"]
    for detection in filtered_detections
)


print("\nOBJECT COUNTS")
print("-" * 60)

for class_name, count in sorted(class_counts.items()):

    print(
        f"  {class_name:<8}: {count}"
    )


# ------------------------------------------------------------
# 8. EXPECTED OBJECT COUNTS
# ------------------------------------------------------------

expected_counts = {
    "bottle": 1,
    "book": 1,
    "bag": 1,
}


assert class_counts == expected_counts, \
    f"Object count mismatch: {class_counts}"


# ------------------------------------------------------------
# 9. BOUNDING BOX VALIDATION
# ------------------------------------------------------------

print("\nBOUNDING BOX VALIDATION")
print("-" * 60)

for detection in filtered_detections:

    x1, y1, x2, y2 = detection["bbox"]

    assert 0 <= x1 < IMAGE_WIDTH
    assert 0 <= x2 <= IMAGE_WIDTH
    assert 0 <= y1 < IMAGE_HEIGHT
    assert 0 <= y2 <= IMAGE_HEIGHT

    assert x2 > x1, \
        "Bounding-box width must be positive"

    assert y2 > y1, \
        "Bounding-box height must be positive"

    print(
        f"  ✅ {detection['class']} "
        f"bounding box valid"
    )


# ------------------------------------------------------------
# 10. DETECTION OUTPUT SCHEMA VALIDATION
# ------------------------------------------------------------

print("\nDETECTION SCHEMA VALIDATION")
print("-" * 60)

required_fields = {
    "class",
    "confidence",
    "bbox",
}

for detection in filtered_detections:

    assert required_fields.issubset(
        detection.keys()
    ), "Detection schema is incomplete"

    assert isinstance(
        detection["class"],
        str
    )

    assert isinstance(
        detection["confidence"],
        float
    )

    assert len(
        detection["bbox"]
    ) == 4

print("✅ Detection schema valid")
print("✅ Class labels valid")
print("✅ Confidence values valid")
print("✅ Bounding boxes valid")


# ------------------------------------------------------------
# 11. FINAL IMAGE ANALYSIS RESULT
# ------------------------------------------------------------

analysis_result = {
    "image_width": IMAGE_WIDTH,
    "image_height": IMAGE_HEIGHT,
    "objects_detected": len(filtered_detections),
    "class_counts": dict(class_counts),
    "detections": filtered_detections,
}


# ------------------------------------------------------------
# 12. FINAL ASSERTIONS
# ------------------------------------------------------------

assert analysis_result["objects_detected"] == 3

assert analysis_result["class_counts"]["bottle"] == 1
assert analysis_result["class_counts"]["book"] == 1
assert analysis_result["class_counts"]["bag"] == 1

assert len(
    analysis_result["detections"]
) == 3


# ------------------------------------------------------------
# FINAL RESULT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("IMAGE ANALYSIS TEST RESULTS")
print("=" * 60)

print("✅ IMAGE CREATION PASSED")
print("✅ IMAGE FORMAT VALIDATION PASSED")
print("✅ DETECTION OUTPUT PASSED")
print("✅ CONFIDENCE FILTERING PASSED")
print("✅ OBJECT COUNTING PASSED")
print("✅ BOUNDING BOX VALIDATION PASSED")
print("✅ DETECTION SCHEMA PASSED")
print("✅ STRUCTURED ANALYSIS OUTPUT PASSED")

print("=" * 60)
print("🎯 IMAGE ANALYSIS TEST PASSED")
print("=" * 60)
```

```python

# ============================================================
# STORE VISION AI — VIDEO + TRACKING TEST
# Synthetic video/tracking validation
# ============================================================

from collections import defaultdict

print("=" * 60)
print("STORE VISION AI — VIDEO + TRACKING TEST")
print("=" * 60)


# ------------------------------------------------------------
# 1. SYNTHETIC VIDEO FRAMES
# ------------------------------------------------------------

# Each record represents one detected customer in a video frame.
#
# zone:
#   ENTRY    -> customer entering
#   STORE    -> customer inside
#   CHECKOUT -> customer at checkout
#   EXIT     -> customer leaving

frames = [
    # C001
    {"frame": 1,  "track_id": "C001", "zone": "ENTRY",    "item": None},
    {"frame": 5,  "track_id": "C001", "zone": "STORE",    "item": "bottle"},
    {"frame": 10, "track_id": "C001", "zone": "STORE",    "item": "bottle"},
    {"frame": 15, "track_id": "C001", "zone": "STORE",    "item": "book"},
    {"frame": 20, "track_id": "C001", "zone": "STORE",    "item": None},
    {"frame": 25, "track_id": "C001", "zone": "CHECKOUT", "item": None},
    {"frame": 30, "track_id": "C001", "zone": "EXIT",     "item": None},

    # C002
    {"frame": 2,  "track_id": "C002", "zone": "ENTRY",    "item": None},
    {"frame": 7,  "track_id": "C002", "zone": "STORE",    "item": "bag"},
    {"frame": 12, "track_id": "C002", "zone": "STORE",    "item": "bag"},
    {"frame": 18, "track_id": "C002", "zone": "STORE",    "item": None},
    {"frame": 22, "track_id": "C002", "zone": "CHECKOUT", "item": None},
    {"frame": 28, "track_id": "C002", "zone": "EXIT",     "item": None},

    # C003
    {"frame": 3,  "track_id": "C003", "zone": "ENTRY",    "item": None},
    {"frame": 8,  "track_id": "C003", "zone": "STORE",    "item": "bottle"},
    {"frame": 12, "track_id": "C003", "zone": "STORE",    "item": "bottle"},
    {"frame": 17, "track_id": "C003", "zone": "STORE",    "item": None},
    {"frame": 20, "track_id": "C003", "zone": "CHECKOUT", "item": None},
    {"frame": 27, "track_id": "C003", "zone": "EXIT",     "item": None},
]


print("\nSYNTHETIC VIDEO")
print("-" * 60)
print(f"Total simulated observations : {len(frames)}")
print(f"Unique tracks                : "
      f"{len(set(x['track_id'] for x in frames))}")

assert len(frames) > 0
assert len(set(x["track_id"] for x in frames)) == 3

print("✅ Synthetic video frames created")


# ------------------------------------------------------------
# 2. TRACK BUILDING
# ------------------------------------------------------------

tracks = defaultdict(list)

for observation in frames:
    tracks[observation["track_id"]].append(observation)


print("\nTRACKING ANALYSIS")
print("-" * 60)

for track_id, observations in sorted(tracks.items()):

    frame_numbers = [
        x["frame"]
        for x in observations
    ]

    print(
        f"{track_id}: "
        f"frames={frame_numbers}"
    )


# ------------------------------------------------------------
# 3. TRACK ID PERSISTENCE
# ------------------------------------------------------------

print("\nTRACK ID PERSISTENCE")
print("-" * 60)

for track_id, observations in sorted(tracks.items()):

    ids = [
        x["track_id"]
        for x in observations
    ]

    assert all(
        current_id == track_id
        for current_id in ids
    )

    print(
        f"  ✅ {track_id} ID remained consistent"
    )


# ------------------------------------------------------------
# 4. ENTRY / EXIT DETECTION
# ------------------------------------------------------------

journeys = {}

for track_id, observations in tracks.items():

    observations = sorted(
        observations,
        key=lambda x: x["frame"]
    )

    entry = next(
        (
            x["frame"]
            for x in observations
            if x["zone"] == "ENTRY"
        ),
        None,
    )

    checkout = next(
        (
            x["frame"]
            for x in observations
            if x["zone"] == "CHECKOUT"
        ),
        None,
    )

    exit_frame = next(
        (
            x["frame"]
            for x in observations
            if x["zone"] == "EXIT"
        ),
        None,
    )

    journeys[track_id] = {
        "entry": entry,
        "checkout": checkout,
        "exit": exit_frame,
    }


print("\nJOURNEY EVENTS")
print("-" * 60)

for track_id, journey in sorted(journeys.items()):

    print(
        f"{track_id}: "
        f"entry={journey['entry']}, "
        f"checkout={journey['checkout']}, "
        f"exit={journey['exit']}"
    )

    assert journey["entry"] is not None
    assert journey["checkout"] is not None
    assert journey["exit"] is not None

    assert (
        journey["entry"]
        < journey["checkout"]
        < journey["exit"]
    )

    print(
        f"  ✅ Entry → Checkout → Exit sequence valid"
    )


# ------------------------------------------------------------
# 5. ITEM ASSOCIATION
# ------------------------------------------------------------

item_associations = defaultdict(list)

for observation in frames:

    if observation["item"] is not None:

        item_associations[
            observation["track_id"]
        ].append(
            {
                "item": observation["item"],
                "frame": observation["frame"],
            }
        )


print("\nITEM-TRACK ASSOCIATION")
print("-" * 60)

for track_id, items in sorted(
    item_associations.items()
):

    print(f"{track_id}:")

    for item in items:

        print(
            f"  {item['item']} "
            f"(frame {item['frame']})"
        )


# ------------------------------------------------------------
# 6. EXPECTED ITEM ASSOCIATIONS
# ------------------------------------------------------------

assert any(
    x["item"] == "bottle"
    for x in item_associations["C001"]
)

assert any(
    x["item"] == "book"
    for x in item_associations["C001"]
)

assert any(
    x["item"] == "bag"
    for x in item_associations["C002"]
)

assert any(
    x["item"] == "bottle"
    for x in item_associations["C003"]
)

print("\n✅ Item-to-track associations validated")


# ------------------------------------------------------------
# 7. VISIT DURATION
# ------------------------------------------------------------

visit_durations = {}

for track_id, journey in journeys.items():

    duration = (
        journey["exit"]
        - journey["entry"]
    )

    visit_durations[track_id] = duration


print("\nVISIT DURATION")
print("-" * 60)

for track_id, duration in sorted(
    visit_durations.items()
):

    print(
        f"{track_id}: "
        f"{duration} frames"
    )

    assert duration > 0


# ------------------------------------------------------------
# 8. TRACK COMPLETENESS
# ------------------------------------------------------------

print("\nTRACK COMPLETENESS")
print("-" * 60)

for track_id in ["C001", "C002", "C003"]:

    assert track_id in tracks
    assert len(tracks[track_id]) >= 3

    print(
        f"  ✅ {track_id} complete track"
    )


# ------------------------------------------------------------
# 9. FINAL STRUCTURED VIDEO RESULT
# ------------------------------------------------------------

video_analysis = {
    "total_frames_observed": len(frames),
    "customers_tracked": len(tracks),
    "journeys": journeys,
    "items": {
        track_id: items
        for track_id, items
        in item_associations.items()
    },
    "visit_durations": visit_durations,
}


assert video_analysis[
    "customers_tracked"
] == 3

assert len(
    video_analysis["journeys"]
) == 3

assert len(
    video_analysis["visit_durations"]
) == 3


# ------------------------------------------------------------
# 10. FINAL RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("VIDEO + TRACKING TEST RESULTS")
print("=" * 60)

print("✅ VIDEO FRAME PROCESSING PASSED")
print("✅ TRACK CREATION PASSED")
print("✅ TRACK ID PERSISTENCE PASSED")
print("✅ ENTRY DETECTION PASSED")
print("✅ CHECKOUT EVENT DETECTION PASSED")
print("✅ EXIT DETECTION PASSED")
print("✅ ITEM-TRACK ASSOCIATION PASSED")
print("✅ VISIT DURATION PASSED")
print("✅ TRACK COMPLETENESS PASSED")
print("✅ STRUCTURED VIDEO ANALYSIS PASSED")

print("=" * 60)
print("🎯 VIDEO + TRACKING TEST PASSED")
print("=" * 60)
```

```python

# ============================================================
# STORE VISION AI — STEP 7
# AI RETAIL STORE ANALYTICS ASSISTANT
# CPU ONLY — NO GPU REQUIRED
# ============================================================

import pandas as pd
import numpy as np
from datetime import datetime

print("Step 7: AI Retail Store Analytics Assistant")
print("Runtime: CPU")
print("Status: Starting...\n")


# ------------------------------------------------------------
# 1. SYNTHETIC STORE DATA
# ------------------------------------------------------------
# Controlled synthetic data for project validation.
# This is NOT real-world customer/store data.

products = pd.DataFrame([
    {
        "product_id": "P001",
        "product": "Notebook",
        "category": "Stationery",
        "stock": 18,
        "reorder_level": 10,
        "unit_cost": 25,
        "selling_price": 40,
        "units_sold": 32
    },
    {
        "product_id": "P002",
        "product": "Bottle",
        "category": "Household",
        "stock": 7,
        "reorder_level": 12,
        "unit_cost": 60,
        "selling_price": 100,
        "units_sold": 45
    },
    {
        "product_id": "P003",
        "product": "Book",
        "category": "Education",
        "stock": 25,
        "reorder_level": 15,
        "unit_cost": 80,
        "selling_price": 130,
        "units_sold": 21
    },
    {
        "product_id": "P004",
        "product": "Pen",
        "category": "Stationery",
        "stock": 55,
        "reorder_level": 20,
        "unit_cost": 8,
        "selling_price": 15,
        "units_sold": 60
    },
    {
        "product_id": "P005",
        "product": "Bag",
        "category": "Accessories",
        "stock": 5,
        "reorder_level": 8,
        "unit_cost": 300,
        "selling_price": 450,
        "units_sold": 14
    },
    {
        "product_id": "P006",
        "product": "Cup",
        "category": "Household",
        "stock": 30,
        "reorder_level": 10,
        "unit_cost": 35,
        "selling_price": 60,
        "units_sold": 10
    }
])

print("Synthetic product dataset created.")
print(products)


# ------------------------------------------------------------
# 2. SALES / REVENUE CALCULATION
# ------------------------------------------------------------

products["revenue"] = (
    products["selling_price"] * products["units_sold"]
)

products["cost"] = (
    products["unit_cost"] * products["units_sold"]
)

products["profit"] = (
    products["revenue"] - products["cost"]
)

products["profit_margin_percent"] = (
    products["profit"] / products["revenue"] * 100
).round(2)


# ------------------------------------------------------------
# 3. TOTAL STORE PERFORMANCE
# ------------------------------------------------------------

total_revenue = products["revenue"].sum()
total_cost = products["cost"].sum()
total_profit = products["profit"].sum()
total_units = products["units_sold"].sum()

overall_margin = (
    total_profit / total_revenue * 100
    if total_revenue > 0 else 0
)

print("\n========== STORE PERFORMANCE ==========")
print(f"Total units sold : {total_units}")
print(f"Total revenue    : ₹{total_revenue:,.2f}")
print(f"Total cost       : ₹{total_cost:,.2f}")
print(f"Total profit     : ₹{total_profit:,.2f}")
print(f"Profit margin    : {overall_margin:.2f}%")


# ------------------------------------------------------------
# 4. TOP SELLING PRODUCTS
# ------------------------------------------------------------

top_products = products.sort_values(
    by="units_sold",
    ascending=False
).head(3)

print("\n========== TOP SELLING PRODUCTS ==========")

for _, row in top_products.iterrows():
    print(
        f"{row['product']} -> "
        f"{row['units_sold']} units sold, "
        f"profit ₹{row['profit']:,.2f}"
    )


# ------------------------------------------------------------
# 5. LOW-STOCK DETECTION
# ------------------------------------------------------------

low_stock = products[
    products["stock"] <= products["reorder_level"]
].copy()

print("\n========== LOW STOCK ALERTS ==========")

if len(low_stock) == 0:
    print("No low-stock products detected.")
else:
    for _, row in low_stock.iterrows():
        print(
            f"ALERT: {row['product']} "
            f"(stock={row['stock']}, "
            f"reorder level={row['reorder_level']})"
        )


# ------------------------------------------------------------
# 6. STOCK VALUE
# ------------------------------------------------------------

products["inventory_value"] = (
    products["stock"] * products["unit_cost"]
)

total_inventory_value = products["inventory_value"].sum()

print("\n========== INVENTORY ==========")
print(
    f"Current inventory value: "
    f"₹{total_inventory_value:,.2f}"
)


# ------------------------------------------------------------
# 7. PRODUCT PERFORMANCE ANALYSIS
# ------------------------------------------------------------

def classify_product(row):

    if row["units_sold"] >= 40 and row["profit_margin_percent"] >= 30:
        return "HIGH_PERFORMER"

    elif row["units_sold"] >= 20:
        return "GOOD_PERFORMER"

    elif row["units_sold"] < 15:
        return "LOW_DEMAND"

    return "MODERATE"


products["performance"] = products.apply(
    classify_product,
    axis=1
)

print("\n========== PRODUCT PERFORMANCE ==========")
print(
    products[
        [
            "product",
            "units_sold",
            "revenue",
            "profit",
            "profit_margin_percent",
            "stock",
            "performance"
        ]
    ].to_string(index=False)
)


# ------------------------------------------------------------
# 8. AI RETAIL RECOMMENDATION ENGINE
# ------------------------------------------------------------

recommendations = []

for _, row in products.iterrows():

    if row["stock"] <= row["reorder_level"]:
        recommendations.append(
            f"Reorder {row['product']} soon. "
            f"Current stock is {row['stock']}."
        )

    if (
        row["units_sold"] >= 40
        and row["profit_margin_percent"] >= 30
    ):
        recommendations.append(
            f"Increase visibility of {row['product']} "
            f"because it has strong sales and margin."
        )

    if row["units_sold"] < 15:
        recommendations.append(
            f"Review demand for {row['product']}. "
            f"Consider promotion, better placement, "
            f"or reducing future stock."
        )


# ------------------------------------------------------------
# 9. CATEGORY ANALYSIS
# ------------------------------------------------------------

category_analysis = products.groupby(
    "category"
).agg(
    revenue=("revenue", "sum"),
    profit=("profit", "sum"),
    units_sold=("units_sold", "sum")
).reset_index()

category_analysis["margin_percent"] = (
    category_analysis["profit"]
    / category_analysis["revenue"]
    * 100
).round(2)

print("\n========== CATEGORY ANALYSIS ==========")
print(category_analysis.to_string(index=False))


# ------------------------------------------------------------
# 10. AUTOMATIC STORE SUMMARY
# ------------------------------------------------------------

best_product = products.loc[
    products["profit"].idxmax()
]

best_sales_product = products.loc[
    products["units_sold"].idxmax()
]

print("\n========== AI STORE SUMMARY ==========")

print(
    f"""
Store performance summary:

• Revenue generated: ₹{total_revenue:,.2f}
• Total cost: ₹{total_cost:,.2f}
• Total profit: ₹{total_profit:,.2f}
• Overall profit margin: {overall_margin:.2f}%
• Total units sold: {total_units}

Highest-profit product:
{best_product['product']} — ₹{best_product['profit']:,.2f}

Highest-sales product:
{best_sales_product['product']} — {best_sales_product['units_sold']} units
"""
)


# ------------------------------------------------------------
# 11. AI RECOMMENDATIONS
# ------------------------------------------------------------

print("========== AI RETAIL RECOMMENDATIONS ==========")

if recommendations:
    for i, recommendation in enumerate(
        recommendations, 1
    ):
        print(f"{i}. {recommendation}")
else:
    print("No major recommendations generated.")


# ------------------------------------------------------------
# 12. CREATE FINAL ANALYTICS REPORT
# ------------------------------------------------------------

analytics_report = {
    "generated_at": datetime.now().isoformat(),
    "total_units_sold": int(total_units),
    "total_revenue": float(total_revenue),
    "total_cost": float(total_cost),
    "total_profit": float(total_profit),
    "overall_margin_percent": float(overall_margin),
    "inventory_value": float(total_inventory_value),
    "low_stock_products": low_stock["product"].tolist(),
    "top_selling_product": best_sales_product["product"],
    "highest_profit_product": best_product["product"],
    "recommendations": recommendations
}

print("\n========== FINAL ANALYTICS REPORT ==========")

for key, value in analytics_report.items():
    print(f"{key}: {value}")


# ------------------------------------------------------------
# 13. SAVE RESULT FOR FUTURE GRADIO INTEGRATION
# ------------------------------------------------------------

products.to_csv(
    "store_analytics_results.csv",
    index=False
)

category_analysis.to_csv(
    "category_analytics.csv",
    index=False
)

print("\nFiles generated:")
print("✓ store_analytics_results.csv")
print("✓ category_analytics.csv")

print("\n============================================")
print("STEP 7 COMPLETED")
print("CPU validation completed successfully.")
print("============================================")
```
